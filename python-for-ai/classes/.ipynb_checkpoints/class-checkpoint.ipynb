### calculating the bonus of employee based on salary

```python

class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def calc_bonus(self):
        return self.salary*0.2
employee=Employee('John',9000)
employee.name
employee.calc_bonus()
```

### finding average of student marks using class

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def find_average(self):
        average=sum(self.marks)/len(self.marks)
        return average
stud1=Student('john',[20,20,20])
stud1.find_average()
```

### finding area of rectangle using class

```python
class Rectangle:
    def __init__(self,width,length):
        self.width=width
        self.length=length
    def calc_area(self):
        area=self.width*self.length
        return area
rectangle=Rectangle(20,20)
rectangle.calc_area()
    
```

### finding the year of human according to today year

```python
from datetime import date
class Human:
    def __init__(self,name,birth_year):
        self.birth_year=birth_year
        self.name=name
    def find_age(self):
        current=date.today().year
        return current- self.birth_year
    def introduce(self):
        print(f"Hello,My name is {self.name} and i am {self.find_age()} years old.")
human=Human("John",2000)
human.introduce()
        
```

### defining class using pydantic base model for structuring response and avoid errors 

```python
from pydantic import BaseModel
from typing import Optional
class User(BaseModel):
    name:str
    age:Optional[int]=None
user=User(name='john')    
print(user)
```
