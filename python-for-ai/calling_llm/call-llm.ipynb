```python
from google import genai
import os
from dotenv import load_dotenv
load_dotenv()
def generate_recipe(ingredients,cuisine,diet):
    '''
    generate a recipe using ingredients provided.
    cuisine can be "Rwandan","chinese","indian","any",etc.
    diet can be "vegan","vegetarian","meat","any",etc.
    '''
    prompt=f'''
    generate one food recipe using these ingredients:{",".join(ingredients)}
    Recipe should not be more thatn 100 words.
    Diet:{diet}
    '''
    try:  
        client = genai.Client(api_key=os.getenv("GOOGLE_API_KEY"))
        response=client.models.generate_content(
            model="gemini-3.5-flash",
            contents=prompt
        )
        if not response.text:
            return "the model returned empty response try again"
        return response.text
    except Exception as e:
        return f"sorry,something went wrong couldnt generate recipe right now.({e})"
    
```

```python
recipe=generate_recipe(["rice","onion","corns","salt"],"indian","any")
print(recipe)
```
