```python
def safe_chai_bill(cups,price_per_cup):
    
    
    try:
        cups=int(cups)
        price_per_cup=float(price_per_cup)
        if cups<1:
            print("Please cups must be at least 1")
            return
        total_bill=cups*price_per_cup    
        return total_bill
    except ValueError as error:    
        print(f"please enter numbers only. {(error)}")
```

```python
safe_chai_bill("whfhfh",40)
```
