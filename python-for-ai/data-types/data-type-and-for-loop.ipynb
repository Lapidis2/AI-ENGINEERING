### hash map in dictionary


```python
business={
    "January":{
        "revenue":2000,
        "expense":100
    },
    "February":{
        "revenue":-3000,
        "expense":150
    },
    "March":{
        "revenue":4000,
        "expense":200
    }
}
```

```python
for month,data in business.items():
    if data['revenue']<0:
        print(f"invalid revenue in {month}.")
        continue
    profit=data["revenue"] - data["expense"]
    margin=profit*100/data["revenue"]
    print(f"{month}:{margin}")
```

# Class exercise student performance tracker


```python
students={
    "eric":[85,90,78],
    "peter":[72,68,75],
    "Nadine":[45,52,48],
    "Nadia":[60,65,70]
}
```

```python
Grade=['Topper','Pass','Needs Improvement']
```

```python
print("===Student averages marks are ===")
for studentName,mark in students.items():
    total=sum(mark)
    average=total/len(mark)
    if average>=80:
        print(f"{studentName}: {average:.2f} {Grade[0]}")
    elif 60<=average <=79:
        print(f"{studentName}: {average:.2f} {Grade[1]}")
    else:    
        print(f"{studentName}: {average:.2f} {Grade[2]}")    
    
```
