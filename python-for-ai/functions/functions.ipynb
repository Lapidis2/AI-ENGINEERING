```python
def find_margin(revenue,expense):
    profit=revenue-expense
    margin=profit*100/revenue
    return margin
   
```

```python
find_margin(100,20) 
```

### returning multiple value like profit and margin


```python
def find_margin_and_profit(revenue:int,expense:int)->float:
    profit=revenue-expense
    margin=profit*100/revenue
    return margin,profit
```

```python
margin,profit=find_margin_and_profit(100,20)
print(f"margin:{margin} and profit:{profit}")
```

### adding argments on function for dictionary

```python
def find_total_expense(**kargs):
    total=0
    for key,value in kargs.items():
        total+=value
    return total
find_total_expense(food=100,rent=200,transport=300,wages=400,cloths=7000)    
```

### function exercises by calculating bill with discount and addons

```python
def chai_bill(cups,*addons,**discount):
    total_bill=cups*15
    for addon in addons:
        total_bill+=5
    if 'discount' in discount:
        total_discount=total_bill*discount["discount"]/100
        total_bill-=total_discount
    return total_bill
print(chai_bill(2))    
print(chai_bill(3,'ginger','milk'))    
print(chai_bill(4,'ginger',discount=10))        
    

```

```python
from pydantic import BaseModel
class YouTubeChanel(BaseModel):
      name:str
      subscribers:int=0
      def subscribe(self):
          self.subscribers+=1
      def unsubscribe(self):
          if self.subscribers>0:
             self.subscribers -=1
      def  show(self):
           print(f" the channel of {self.name} has {self.subscribers} Subscribers ")
channel1=YouTubeChanel(name="Jean Pierre")
channel2=YouTubeChanel(name="Pierre")
channel1.subscribe()
channel1.subscribe()
channel2.subscribe()
channel2.unsubscribe()

channel1.show()
channel2.show()

        
```
